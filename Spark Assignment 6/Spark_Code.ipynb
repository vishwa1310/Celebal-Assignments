```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import col, when, count,avg, min, max
```

```python
spark = SparkSession.builder \
    .master("local[*]") \
    .appName("Assignment") \
    .config("spark.hadoop.io.native.lib.available", "false") \
    .getOrCreate()
print(spark.version)
```

```python
df_file= (
    spark.read
    .option("header", True)
    .option("nullValue", "null")
    .option("inferSchema", True)
    .csv("C:\\Users\\khand\\Downloads\\Grocery_Inventory_and_Sales_Dataset.csv")
)
```

```python
df_file.show()
```

```python
df_file.printSchema()
```

```python
df_file.columns
```

```python
df_file.filter(col("Catagory") == "Dairy").show()
```

```python
df_file.filter(col("Stock_Quantity") > 50).show()
```

```python
df_file.filter(col("Status") == "Active").show()
```

```python
df_file.select( "Product_Name", "Stock_Quantity", "Unit_Price","Status").show()

```

```python
df_new = (
    df_file.withColumnRenamed("Unit_Price", "Price")
        .withColumnRenamed("Stock_Quantity", "Available stock")
        .withColumnRenamed("Expiration_Date", "Expiry_Date"))
```

```python
df_new.columns
```

```python
from pyspark.sql.functions import regexp_replace

df_new = df_new.withColumn(
    "Price",
    regexp_replace(col("Price"), "\\$", "").cast("double")
)
```

```python
df_new.printSchema()
```

```python
df_new = df_new.withColumn("Inventory_Value", col("Price") * col("Available stock"))
```

```python
df_new.select( "Product_Name", "Price","Available stock","Inventory_Value").show()
```

```python
df_new.count()
```

```python
df_new.groupBy("Catagory") \
      .agg(F.sum("Sales_Volume").alias("Total Sales")) \
      .show()
```

```python
df_new.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df_new.columns]).show()
```

```python
df_new.filter(F.col("Catagory").isNull()).show()
```

```python
df_new = df_new.filter(F.col("Catagory").isNotNull())
```

```python
df_new.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df_new.columns
]).show()
```

```python
df_new.select(F.max("Inventory_Value").alias("Highest_Inventory")).show()
```

```python
df_new.filter(F.col("Status") == "Active").select( "Product_Name", "Catagory","Price", "Inventory_Value").show()
```

```python
lazy_chain = (
    df_new
    .filter(F.col("Price") > 5)
    .select("Product_Name","Catagory", "Price"  ))
```

```python
lazy_chain.show()
```
