```sql
SELECT COUNT(*) AS Total_CRM_Records
FROM workspace.default.crm_silver;
```

```sql
SELECT COUNT(*) AS Total_Billing_Records
FROM workspace.default.billing_silver;
```

```sql
SELECT
status,
COUNT(*) AS Total
FROM workspace.default.billing_silver
GROUP BY status
ORDER BY Total DESC;
```

```sql
SELECT
city,
COUNT(*) AS Customers
FROM workspace.default.crm_silver
GROUP BY city
ORDER BY Customers DESC;
```

```sql
SELECT
DATE_FORMAT(transaction_date,'yyyy-MM') AS Month,
SUM(amount) AS Revenue
FROM workspace.default.billing_silver
GROUP BY DATE_FORMAT(transaction_date,'yyyy-MM')
ORDER BY Month;
```

```sql
SELECT
DATE_FORMAT(transaction_date,'yyyy-MM') AS Month,
COUNT(*) AS Transactions
FROM workspace.default.billing_silver
GROUP BY DATE_FORMAT(transaction_date,'yyyy-MM')
ORDER BY Month;
```

```sql
SELECT
ROUND(AVG(amount),2) AS Average_Transaction
FROM workspace.default.billing_silver;
```

```sql
SELECT
status,
ROUND(SUM(amount),2) AS Revenue
FROM workspace.default.billing_silver
GROUP BY status
ORDER BY Revenue DESC;
```

```sql
SELECT
transaction_date,
SUM(amount) AS Revenue
FROM workspace.default.billing_silver
GROUP BY transaction_date
ORDER BY transaction_date;
```
