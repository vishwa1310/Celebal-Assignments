```python
crm_df = spark.table("workspace.default.crm_bronze")

billing_df = spark.table("workspace.default.billing_bronze")

analytics_df = spark.table("workspace.default.analytics_bronze")
```

```python
display(crm_df)

display(billing_df)

display(analytics_df)
```

```python
print("CRM Records :", crm_df.count())

print("Billing Records :", billing_df.count())

print("Analytics Records :", analytics_df.count())
```

```python
from pyspark.sql.functions import *

crm_df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in crm_df.columns
]).show()
```

```python
billing_df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in billing_df.columns
]).show()
```

```python
analytics_df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in analytics_df.columns
]).show()
```

```python
print(
    "CRM Duplicate Records:",
    crm_df.count() - crm_df.dropDuplicates().count()
)
print(
    "Billing Duplicate Records:",
    billing_df.count() - billing_df.dropDuplicates().count()
)
print(
    "Analytics Duplicate Records:",
    analytics_df.count() - analytics_df.dropDuplicates().count()
)
```

```python
crm_clean = (
    crm_df
    .dropDuplicates()
    .dropna()
)
```

```python
billing_clean = (
    billing_df
    .dropDuplicates()
    .dropna()
)
```

```python
analytics_clean = (
    analytics_df
    .dropDuplicates()
    .dropna()
)
```

```python
print("CRM :", crm_clean.count())

print("Billing :", billing_clean.count())

print("Analytics :", analytics_clean.count())
```

```python
crm_clean.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("workspace.default.crm_silver")

billing_clean.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("workspace.default.billing_silver")

analytics_clean.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("workspace.default.analytics_silver")
```

```python
crm_silver = spark.table("workspace.default.crm_silver")

billing_silver = spark.table("workspace.default.billing_silver")

analytics_silver = spark.table("workspace.default.analytics_silver")
```
